```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import wbdata

pd.set_option("display.max_colwidth", 90)
pd.set_option("display.max_rows", 200)
sns.set_theme(style="whitegrid")

DATA_DIR = Path("..") / "data"
wdi = pd.read_csv(DATA_DIR / "sdg_wdi" / "WDICSV.csv")
series = pd.read_csv(DATA_DIR / "sdg_wdi" / "WDISeries.csv")
country = pd.read_csv(DATA_DIR / "sdg_wdi" / "WDICountry.csv")
print(wdi.shape, series.shape, country.shape)
print(series.columns.tolist())
```

```python
for name, df in {"wdi": wdi, "series": series, "country": country}.items():
    non_year = []
    for c in df.columns:
        if not str(c).strip().isdigit():
            non_year.append(c)                  # excludes the column names for years to keep the list short.
    print(name, df.shape)                       
    print(non_year, "\n")

print("wdi & series:    ", sorted(set(wdi.columns) & set(series.columns)))
print("wdi & country:   ", sorted(set(wdi.columns) & set(country.columns)))
print("series & country:", sorted(set(series.columns) & set(country.columns)))
```
